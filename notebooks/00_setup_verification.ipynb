# Extending Image Recognition Attacks by Embedding Aggregation over Time

This project contains the experiment for the master thesis _Extending Image Recognition Attacks by Embedding Aggregation over Time_. It consists of a Python library called _meles_, a workflow experiment using _SnakeMake_, and a set of notebooks for setting the experiment up and visualizing the workflow experiment results.

### Getting started

The project uses [uv](https://docs.astral.sh/uv/getting-started/installation/) to manage its environment and dependencies. Install it by going to the official website or running the following (might be deprecated).

```shell
# Windows
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"

# macOS and Linux
curl -LsSf https://astral.sh/uv/install.sh | sh
```
